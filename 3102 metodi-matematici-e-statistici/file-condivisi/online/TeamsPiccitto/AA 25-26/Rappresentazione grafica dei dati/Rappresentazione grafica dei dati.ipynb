```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

Istogramma 1

```python
mod = ["automobile","a piedi","autobus","scooter","bicicletta"]
freq = np.array([7, 3, 9, 4, 5])

# crea istogramma
fig, ax = plt.subplots(1, 1)
plt.bar(mod, freq/np.sum(freq))

# set titolo
ax.set_title("Mezzi di trasporto utilizzati")
  
# set assi
ax.set_xlabel('modalità')
ax.set_ylabel('frequenze')

plt.show()
```

Rappresentazione a torta

```python
# crea istogramma
fig, ax = plt.subplots(1, 1)
ax.pie(freq, labels=mod, autopct='%1.1f%%')

# set titolo
ax.set_title("Mezzi di trasporto utilizzati")

plt.show()
```

Istogramma 2

```python
X = np.loadtxt('Data_altezze.dat')
```

```python
N = np.size(X)
m = np.sum(X)/N
s = np.sum((X-m)**2)/N
d = np.sqrt(s)
print(m, s, d)
```

```python
m2 = np.mean(X)
s2 = np.var(X)
d2 = np.std(X)
print(m2, s2, d2)
```

```python
# Costruire un istogramma a 20 barre

fig, ax = plt.subplots(1, 1)
# ax.hist(X)
ax.hist(X, bins=20, edgecolor = "black")

ax.set_title("Indagine sull'altezza di una popolazione")
ax.set_xlabel('altezza (cm)')
ax.set_ylabel('frequenze')

plt.show()
```

```python
mu3 = np.sum((X-m)**3)/N
mu4 = np.sum((X-m)**4)/N

sk = mu3/d**3
kur = mu4/s**2

print(sk, kur)
```

```python
from scipy.stats import skew
from scipy.stats import kurtosis
sk2 = skew(X)
kur2 = kurtosis(X, fisher=False)
print(sk2, kur2)
```

```python
xx = np.linspace(130, 210, num=200)
yy = norm.pdf(xx, m, d)
fig, ax = plt.subplots(1, 1)
# ax.hist(X)
ax.hist(X, density=True, bins=20, edgecolor = "black")

ax.set_title("Indagine sull'altezza di una popolazione")
ax.set_xlabel('altezza (cm)')
ax.set_ylabel('frequenze')

plt.plot(xx,yy)
plt.show()
```

Box plot: la scatola è alta IQR = q3-q1, la linea arancione è la mediana, i baffi finiscono a 1.5*IQR, gli altri dati sono outlier.

```python
# crea box_plot
fig, ax = plt.subplots(1, 1)

ax.boxplot(X)
plt.show()
```

Grafici quantile-quantile

```python
from scipy.stats import probplot, expon
fig, ax = plt.subplots(1, 1)
probplot(X, dist=norm, plot=ax)
plt.show()
```

```python
Y = np.random.exponential(scale=2,size=500)
fig, ax = plt.subplots(1, 1)
probplot(Y, dist=norm, plot=ax)
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
probplot(Y, dist=expon, plot=ax)
plt.show()
```
