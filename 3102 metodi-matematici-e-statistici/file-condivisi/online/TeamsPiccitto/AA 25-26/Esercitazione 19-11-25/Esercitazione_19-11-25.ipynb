```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
pesi = np.array(np.arange(15,81,5)) 
mm = np.array([0, 1, 2, 20, 34, 50, 64, 59, 40, 18, 11, 1, 0, 0])
wl = np.array([1, 6, 19, 23, 52, 75, 82, 93, 63, 44, 21, 10, 8, 3])
sl = 3
sm = 5
alpha = 0.05
n_mm = np.sum(mm)
n_wl = np.sum(wl)
print(n_mm, n_wl)
```

```python
plt.bar(pesi, mm/n_mm, width=2.5)
plt.bar(pesi, wl/n_wl, width=2.0)
```

```python
media_mm = np.sum(pesi*mm)/n_mm
media_wl = np.sum(pesi*wl)/n_wl

print(media_mm, media_wl)
```

```python
err_mm = sm*norm.ppf(1-alpha*0.5)/np.sqrt(n_mm)
err_wl = sl*norm.ppf(1-alpha*0.5)/np.sqrt(n_wl)

Im_left = media_mm - err_mm
Im_right = media_mm + err_mm

Il_left = media_wl - err_wl
Il_right = media_wl + err_wl

print(Im_left, Im_right)
print(Il_left, Il_right)
```

```python
Z_n = (media_mm- media_wl)/(np.sqrt(sm/n_mm + sl/n_wl))
print(Z_n)
print(norm.ppf(alpha))
```

Dato che il la mia variabile Zn è più piccola del quartile di ordine alpha rifiuto l'ipotesi nulla in favore di quella alternativa. In alternativa posso valutare il p-value e confrontarlo con alpha

```python
pvalue = norm.cdf(Z_n)
print(pvalue)
```

Dato che alpha è maggiore del mio pvalue, rifiuto l'ipotesi nulla in favore di quella alternativa!
