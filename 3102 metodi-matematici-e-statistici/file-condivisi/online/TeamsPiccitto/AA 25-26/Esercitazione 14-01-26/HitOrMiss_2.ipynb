```python
a = 0
b = np.pi/2
M = 1
n = 1000
```

```python
x1 = np.random.rand(n)
csi = a + (b-a)*x1

x2 = np.random.rand(n)
eta = M*x2


y = np.where(eta <= np.cos(csi), csi, 'elimina')
y = y[y != 'elimina']
#print(y)
y = np.array(y, dtype=float)
#print(y)
punti_scartati = n - np.size(y)
print(punti_scartati)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(8, 7))

yr = np.linspace(0, np.pi/2, 500)

ax.hist(y, 50, density=True)
ax.plot(x, np.cos(x), 'r-')
ax.set_xlabel('y1', fontsize=16)
ax.set_ylabel('frequenza', fontsize=16)
ax.set_xlim(0, np.pi/2)
ax.set_xticks(range(0, 3))

plt.show()
```
