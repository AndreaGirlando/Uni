```python
from scipy.stats import binom, geom
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Genero il lancio della moneta
n = 10 #numero di monete
p = 0.2 #probabilità
N = 2000 #numero di volte che ripeto l'esperimento

#prove = np.random.randint(0, 2, (N, n)) 
prove = np.random.rand(N, n)
print(prove)
prove = np.where(prove < p, 1, 0)
print(prove)
print(prove.shape)
testa = np.sum(prove, axis=1) 
print(testa.shape)
print(testa)

F = np.unique(testa, return_counts=True)
print(F)
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(10, 8))
ax.set_xticks(np.arange(0, n+1))
ax.set_title("n=200")
ax.set_xlabel('x', fontsize=16)
ax.set_ylabel('frequenza', fontsize=12)
ax.bar(F[0], F[1]/N, width=0.5, align='center')
ax.plot(F[0], binom.pmf(F[0], n, p), 'or')
plt.show()
```
