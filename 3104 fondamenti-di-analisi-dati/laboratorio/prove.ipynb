```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

data = pd.read_csv("data.csv")

formattedData = data.drop("Student_ID", axis=1).drop("Gender", axis=1).drop("Education_Level", axis=1)
```

```python
sns.heatmap(formattedData.corr(), annot=True)
plt.show()
```
